```python
import json

with open("raw_tables.json", encoding="utf-8") as f:
    tables = json.load(f)

print("Total tables:", len(tables))
print("First table metadata:", {
    k: v for k, v in tables[0].items() if k != "data"
})
print("First table contents:")
for row in tables[0]["data"][:5]:
    print(row)
```

```python
target_pages = [14, 15, 16, 17, 18, 19, 21]

for table in tables:
    if table["pdf_page"] in target_pages:
        print("\n" + "=" * 70)
        print(
            f"PDF Page: {table['pdf_page']} | "
            f"Table: {table['table_number']} | "
            f"Rows: {table['row_count']} | "
            f"Columns: {table['column_count']}"
        )

        for row in table["data"][:8]:
            print(row)
```
